## 5. Menghapus Kolom yang Tidak Diperlukan

Kolom `Id` tidak digunakan dalam proses pemodelan karena hanya berfungsi sebagai identitas data, sehingga perlu dihapus.

```python
train = train.drop('Id', axis=1)
```

## 6. Pemisahan Fitur dan Target

Dataset dipisahkan menjadi:
- X (fitur) → variabel independen
- y (target) → variabel yang akan diprediksi yaitu quality

```python
X = train.drop('quality', axis=1)
y = train['quality']
```

## 7. Normalisasi Data (Feature Scaling)

Dilakukan normalisasi menggunakan StandardScaler agar setiap fitur memiliki skala yang sama, sehingga model dapat bekerja lebih optimal.

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
```

### Interpretasi Preprocessing

Pada tahap ini dilakukan beberapa proses persiapan data sebelum pemodelan, yaitu:

1. Kolom `Id` dihapus karena tidak berpengaruh terhadap prediksi.
2. Dataset dipisahkan menjadi fitur (X) dan target (y), dimana target yang digunakan adalah `quality`.
3. Dilakukan normalisasi data menggunakan StandardScaler sehingga setiap fitur memiliki skala yang sama.

Hasil normalisasi menunjukkan bahwa nilai data telah berubah menjadi memiliki rata-rata sekitar 0 dan standar deviasi 1.

## 8. Pembagian Data Training dan Testing

Data dibagi menjadi data training dan data testing untuk mengevaluasi performa model.

```python
from sklearn.model_selection import train_test_split

X_train, X_val, y_train, y_val = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42
)
```

```python
print(X_train.shape)
print(X_val.shape)
```

### Interpretasi Pembagian Data

Berdasarkan hasil pembagian data, diperoleh:

- Data training (X_train): 685 data dengan 11 fitur
- Data validation/testing (X_val): 172 data dengan 11 fitur

Pembagian dilakukan dengan proporsi 80% data training dan 20% data testing.

Data training digunakan untuk melatih model, sedangkan data testing digunakan untuk mengevaluasi performa model.

Jumlah fitur sebanyak 11 menunjukkan bahwa semua variabel independen telah digunakan setelah menghapus kolom `Id` dan memisahkan variabel target `quality`.

Pembagian ini sudah sesuai dan cukup untuk membangun model klasifikasi yang baik.

## 9. Pembuatan Model Klasifikasi

Pada tahap ini digunakan algoritma Random Forest untuk membangun model klasifikasi dalam memprediksi kualitas anggur.

```python
from sklearn.ensemble import RandomForestClassifier

model = RandomForestClassifier(random_state=42)

model.fit(X_train, y_train)
```

## 10. Prediksi Data Validasi

Model yang telah dilatih digunakan untuk memprediksi data validation.

```python
y_pred = model.predict(X_val)
```

## 11. Evaluasi Model

Evaluasi dilakukan untuk mengetahui performa model menggunakan metrik akurasi dan confusion matrix.

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

print("Akurasi:", accuracy_score(y_val, y_pred))
print("Confusion Matrix:\n", confusion_matrix(y_val, y_pred))
```

### Interpretasi Hasil Evaluasi Model

Berdasarkan hasil evaluasi model Random Forest, diperoleh:

1. Akurasi Model  
Nilai akurasi sebesar 0.5814 atau sekitar 58.14%.  
Hal ini menunjukkan bahwa model mampu memprediksi dengan benar sekitar 58% dari total data.

Akurasi ini tergolong sedang, yang berarti model sudah mampu menangkap pola dalam data, namun masih terdapat kesalahan prediksi yang cukup signifikan.

2. Confusion Matrix  

Confusion matrix menunjukkan distribusi hasil prediksi model terhadap nilai sebenarnya.

- Pada kelas kualitas 5 dan 6, model cukup baik dalam melakukan prediksi, terlihat dari jumlah prediksi benar yang relatif tinggi.
- Namun, pada kelas kualitas yang lebih rendah atau lebih tinggi (seperti 3, 4, dan 7), model kurang mampu memprediksi dengan baik.
- Terlihat banyak kesalahan prediksi antar kelas yang berdekatan, misalnya antara kualitas 5 dan 6, yang menunjukkan bahwa karakteristik data antar kelas tersebut cukup mirip.

3. Analisis Kesalahan Model  
Model cenderung:
- Sulit membedakan kelas dengan jumlah data sedikit (kelas minoritas)
- Lebih sering memprediksi ke kelas tengah (5 dan 6)

Hal ini kemungkinan disebabkan oleh distribusi data yang tidak seimbang (imbalanced dataset).

Kesimpulan:
Model Random Forest yang digunakan sudah mampu melakukan prediksi, namun performanya masih dapat ditingkatkan, terutama dalam membedakan kelas dengan nilai kualitas yang jarang muncul.
